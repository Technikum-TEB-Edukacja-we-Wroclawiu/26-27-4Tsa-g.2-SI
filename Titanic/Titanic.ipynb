```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv('titanic.csv')
data
```

```python
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare']
X = data[features]
y = data['Survived']
```

```python
X['Sex'] = X['Sex'].map({
    'male': 0,
    'female': 1
})

X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1410, stratify = y)
```

```python
mediana = X_train['Age'].median()
X_train = X_train.fillna({"Age": mediana})
X_test = X_test.fillna({"Age": mediana})
```

```python
X_train
```

```python
model = DecisionTreeClassifier(max_depth=3, random_state=1410)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
print(f'Dokładność {round(accuracy_score(y_test, y_pred), 3)}')
```

```python
plt.figure(figsize=(15, 8))

plot_tree(
    model,
    feature_names=features,
    class_names=["nie przeżył", 'przeżył'],
    filled=True,
    rounded=True,
    fontsize=10
)

plt.title('Drzewo decyzyjne')
plt.show()
```
